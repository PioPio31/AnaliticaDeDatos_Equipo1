En esta actividad usarás el archivo de datos `Forbes_2000_top_company.csv`. Este conjunto contiene las 2000 principales empresas, clasificadas mediante el uso de cuatro métricas: ventas, ganancias, activos y valor de mercado. Tiene las 11 columnas de datos:

# 1. Gráfica de Barras de Datos Categóricos
## Gráfica de barras para 'Organization Name'

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Top 10 empresas con mayores ganancias
top_10 = forbes_df.nlargest(10, 'Profits (Billions)')

plt.figure(figsize=(10, 6))
plt.barh(top_10['Organization Name'], top_10['Profits (Billions)'], color='steelblue')
plt.title('Top 10 Empresas por Ganancias')
plt.xlabel('Ganancias (Miles de Millones)')
plt.ylabel('Empresa')
plt.gca().invert_yaxis()  #Para que la mayor quede arriba
plt.tight_layout()
plt.show()
```

## Gráfica de barras para 'Industry'

```python
# Gráfica de barras para las 10 industrias más frecuentes
plt.figure(figsize=(12, 7))
forbes_df['Industry'].value_counts().head(10).plot(kind='bar', color='skyblue')
plt.title('Top 10 Industrias por Número de Empresas')
plt.xlabel('Industria')
plt.ylabel('Número de Empresas')
plt.xticks(rotation=45, ha='right') # Rotar las etiquetas para que se lean mejor
plt.tight_layout() # Ajusta el gráfico para que no se corten las etiquetas
plt.show()
```

## Gráfica de barras para 'Country'

```python
# Gráfica de barras para los 10 países más frecuentes
plt.figure(figsize=(12, 7))
forbes_df['Country'].value_counts().head(10).plot(kind='bar', color='lightcoral')
plt.title('Top 10 Países por Número de Empresas')
plt.xlabel('País')
plt.ylabel('Número de Empresas')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## Gráfica de barras para 'CEO'

```python
# Top 10 CEO por ganancias de su empresa
top_ceos = forbes_df.nlargest(10, 'Profits (Billions)')

plt.figure(figsize=(10, 6))
plt.barh(top_ceos['CEO'], top_ceos['Profits (Billions)'], color='teal')
plt.title('Top 10 CEO por Ganancias de su Empresa')
plt.xlabel('Ganancias (Miles de Millones)')
plt.ylabel('CEO')
plt.gca().invert_yaxis()  # El mayor queda arriba
plt.tight_layout()
plt.show()
```

# 3. Diagrama de Dispersión (Scatter Plot)
## Dispersión Ingresos vs Ganancias

```python
plt.figure(figsize=(10, 6))
plt.scatter(forbes_df['Revenue (Billions)'],
            forbes_df['Profits (Billions)'],
            alpha=0.5, color='green')
plt.title('Ingresos vs Ganancias')
plt.xlabel('Ingresos (Miles de Millones)')
plt.ylabel('Ganancias (Miles de Millones)')
plt.show()
```

## Dispersión Assets vs Market Value

```python
plt.figure(figsize=(10, 6))
plt.scatter(forbes_df['Assets (Billions)'],
            forbes_df['Market Value (Billions)'],
            alpha=0.5, color='purple')
plt.title('Activos vs Valor de Mercado')
plt.xlabel('Activos (Miles de Millones)')
plt.ylabel('Valor de Mercado (Miles de Millones)')
plt.show()
```

# 4. Mapa de Calor (Heatmap)

```python
import seaborn as sns

# Matriz de correlación
corr = forbes_df[['Revenue (Billions)', 'Profits (Billions)',
                  'Assets (Billions)', 'Market Value (Billions)',
                  'Total Employees']].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='Blues', fmt='.2f')
plt.title('Mapa de Calor - Correlación entre Variables')
plt.show()
```
