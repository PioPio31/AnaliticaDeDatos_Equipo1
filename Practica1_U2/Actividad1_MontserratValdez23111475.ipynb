# **Tarea. Visualizacion y exploracion de datos**

```python
#Libreras necesarias para las graficas
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

**Gráficas de barras de los datos categóricos**

```python
#Identificar las cuatro variables de tipo object
categoricas = forbes_df.select_dtypes(include='object').columns

print(categoricas)
```

```python
#Graficas de barras de las variables categoricas
# Gráficas de barras de las variables categóricas
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.flatten()

for i, col in enumerate(categoricas):
    conteo = forbes_df[col].value_counts().head(10)

    sns.barplot(
        x=conteo.index,
        y=conteo.values,
        ax=axes[i]
    )

    axes[i].set_title(f'Conteo de {col}', fontsize=12)
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frecuencia')

    axes[i].tick_params(axis='x', labelrotation=45)

    for etiqueta in axes[i].get_xticklabels():
        etiqueta.set_horizontalalignment('right')

plt.tight_layout()
plt.show()
```

**Boxplot de todos los datos numéricos**

```python
# Identificar las variables numéricas
numericas = forbes_df.select_dtypes(include=np.number).columns

print("Variables numéricas:")
print(numericas)
```

```python
# Boxplots de todas las variables numéricas
fig, axes = plt.subplots(3, 2, figsize=(13, 12))
axes = axes.flatten()

for i, columna in enumerate(numericas):
    sns.boxplot(x=forbes_df[columna], ax=axes[i], showmeans=True)
    axes[i].set_title("Boxplot - " + columna)
    axes[i].set_xlabel(columna)

plt.tight_layout()
plt.show()
```

**Detección de datos atípicos**

```python
# Detectar valores atípicos mediante el método IQR
resultado_atipicos = []

for columna in numericas:
    Q1 = forbes_df[columna].quantile(0.25)
    Q3 = forbes_df[columna].quantile(0.75)
    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    atipicos = forbes_df[
        (forbes_df[columna] < limite_inferior) |
        (forbes_df[columna] > limite_superior)
    ]

    resultado_atipicos.append({
        "Variable": columna,
        "Cantidad de atípicos": len(atipicos),
        "¿Existen atípicos?": "Sí" if len(atipicos) > 0 else "No"
    })

pd.DataFrame(resultado_atipicos)
```

**Diagrama de dispersión**

```python
# Diagrama de dispersión entre ganancias y valor de mercado
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=forbes_df,
    x='Profits (Billions)',
    y='Market Value (Billions)'
)

plt.title('Ganancias vs. valor de mercado')
plt.xlabel('Ganancias (Billions)')
plt.ylabel('Valor de mercado (Billions)')
plt.show()
```

```python
# Correlación entre las variables seleccionadas
forbes_df[['Profits (Billions)', 'Market Value (Billions)']].corr()
```

**Mapa de calor**

```python
# Matriz de correlación de todas las variables numéricas
correlacion = forbes_df.corr(numeric_only=True)

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlacion,
    annot=True,
    fmt='.2f'
)

plt.title('Mapa de calor de correlación')
plt.show()
```
