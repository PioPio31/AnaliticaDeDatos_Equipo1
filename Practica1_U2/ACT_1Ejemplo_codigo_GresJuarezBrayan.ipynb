2. Boxplot de todos los datos numéricos

```python
datos_numericos = forbes_df.select_dtypes(include='number')
datos_numericos.columns
```

```python
datos_numericos = forbes_df.select_dtypes(include='number')

datos_numericos.plot(
    kind='box',
    figsize=(15,7),
    subplots=True,
    layout=(2,3),
    title='Boxplot de las variables numéricas'
)
```

```python
# 3. Detectar datos atípicos
```

```python
for columna in datos_numericos.columns:

    Q1 = datos_numericos[columna].quantile(0.25)
    Q3 = datos_numericos[columna].quantile(0.75)

    IQR = Q3 - Q1

    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR

    atipicos = datos_numericos[
        (datos_numericos[columna] < limite_inferior) |
        (datos_numericos[columna] > limite_superior)
    ]

    print(columna, ':', len(atipicos), 'datos atípicos')
```
