```python
numericas = credit_df.select_dtypes(include = np.number)
numericas.describe()
```

```python
numericas.skew()
```

```python
asimetria_person_age = credit_df['person_age'].skew()

if asimetria_person_age > 1:
    print("La asimetría de person_age es positiva")

elif asimetria_person_age < -1:
    print("La asimetría de person_age es negativa")

else:
    print("La distribución de person_age es aproximadamente simétrica")
```

```python
kurtosis_person_age = credit_df['person_age'].kurt()

if kurtosis_person_age > 0:
    print("La curtosis de person_age es leptocúrtica")

elif kurtosis_person_age < 0:
    print("La curtosis de person_age es platicúrtica")

else:
    print("La curtosis de person_age es mesocúrtica")
```

```python
asimetria_loan_int_rate = credit_df['loan_int_rate'].skew()

if asimetria_loan_int_rate > 1:
    print("La asimetría de loan_int_rate es positiva")

elif asimetria_loan_int_rate < -1:
    print("La asimetría de loan_int_rate es negativa")

else:
    print("La distribución de loan_int_rate es aproximadamente simétrica")
```

```python
kurtosis_loan_int_rate = credit_df['loan_int_rate'].kurt()

if kurtosis_loan_int_rate > 0:
    print("La curtosis de loan_int_rate es leptocúrtica")

elif kurtosis_loan_int_rate < 0:
    print("La curtosis de loan_int_rate es platicúrtica")

else:
    print("La curtosis de loan_int_rate es mesocúrtica")
```

*   Como podrás observar hay valores atípicos en todas las variables. Ejecuta el siguiente código para identificar los valores atípicos en la variable `person_age`

**Interpretación Resultados**

<FONT COLOR="ORANGE">AGREGA AQUI LA INTERPRETACION DE TUS RESULTADOS</FONT>

```python
#COMENTE CADA LINEA DE CODIGO EXPLICANDO SU USO
# Calcula el primer cuartil (25%) de las edades
percentile_25 = credit_df["person_age"].quantile(0.25)
# Calcula el tercer cuartil (75%) de las edades
percentile_75 = credit_df["person_age"].quantile(0.75)
# Calcula el rango intercuartil (IQR) restando Q1 a Q3
iqr = percentile_75 - percentile_25
# Calcula el limite superior para detectar valores atipicos
upper_limit = percentile_75 + 1.5 * iqr
# Calcula el limite inferior para detectar valores atipicos
lower_limit = percentile_25 - 1.5 * iqr
# Filtra las edades que estan fuera de los limites establecidos
IQR_outliers = credit_df[(credit_df["person_age"] < lower_limit) | (credit_df["person_age"] > upper_limit)]
# Muestra los registros que contienen edades atipicas
IQR_outliers
```

```python
#CODIGO PARA MAPA DE CALOR
plt.figure(figsize=(10, 8))

sns.heatmap(credit_df.corr(numeric_only=True), annot=True)

plt.title("Mapa de calor de correlación")

plt.show()
```

Sin embargo, la correlación es moderada y no implica necesariamente una relación causal.
